```python
missing_analysis = pd.DataFrame({
    'Missing Values': df.isnull().sum(),
    'Missing Percentage': (
        df.isnull().sum() / len(df) * 100
    )
})

missing_analysis = missing_analysis.sort_values(
    by='Missing Percentage',
    ascending=False
)

missing_analysis
```

```python
categorical_cols = [
    'Category',
    'Sub-Category',
    'Region',
    'Segment',
    'Ship Mode'
]

for col in categorical_cols:
    print("\n", "="*40)
    print("Column:", col)
    print(df[col].value_counts())
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Ship Date'] = pd.to_datetime(df['Ship Date'])

print("Minimum Order Date:", df['Order Date'].min())
print("Maximum Order Date:", df['Order Date'].max())

print("Minimum Ship Date:", df['Ship Date'].min())
print("Maximum Ship Date:", df['Ship Date'].max())
```

```python
df['Year'] = df['Order Date'].dt.year
df['Month'] = df['Order Date'].dt.month
df['Month Name'] = df['Order Date'].dt.month_name()
df['Day'] = df['Order Date'].dt.day_name()

df[['Order Date', 'Year', 'Month', 'Month Name', 'Day']].head()
```

```python
outlier_results = []

for col in ['Sales', 'Profit', 'Quantity', 'Discount']:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_limit) |
        (df[col] > upper_limit)
    ]

    outlier_results.append({
        'Column': col,
        'Lower Limit': lower_limit,
        'Upper Limit': upper_limit,
        'Outlier Count': len(outliers)
    })

outlier_df = pd.DataFrame(outlier_results)

outlier_df
```

```python
correlation = df[
    ['Sales', 'Profit', 'Quantity', 'Discount']
].corr()

correlation
```

```python
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Heatmap')
plt.show()
```

```python
total_sales = df['Sales'].sum()
total_profit = df['Profit'].sum()
total_orders = df['Order ID'].nunique()
total_customers = df['Customer ID'].nunique()
total_products = df['Product ID'].nunique()
average_discount = df['Discount'].mean()

print("Total Sales:", round(total_sales, 2))
print("Total Profit:", round(total_profit, 2))
print("Total Orders:", total_orders)
print("Total Customers:", total_customers)
print("Total Products:", total_products)
print("Average Discount:", round(average_discount, 2))
```

```python
category_analysis = df.groupby('Category').agg(
    Total_Sales=('Sales', 'sum'),
    Total_Profit=('Profit', 'sum'),
    Total_Quantity=('Quantity', 'sum')
).sort_values(by='Total_Sales', ascending=False)

category_analysis
```

```python
region_analysis = df.groupby('Region').agg(
    Total_Sales=('Sales', 'sum'),
    Total_Profit=('Profit', 'sum'),
    Total_Orders=('Order ID', 'nunique')
).sort_values(by='Total_Sales', ascending=False)

region_analysis
```

```python
segment_analysis = df.groupby('Segment').agg(
    Total_Sales=('Sales', 'sum'),
    Total_Profit=('Profit', 'sum'),
    Total_Orders=('Order ID', 'nunique')
).sort_values(by='Total_Sales', ascending=False)

segment_analysis
```

```python
shipping_analysis = df.groupby('Ship Mode').agg(
    Total_Sales=('Sales', 'sum'),
    Total_Orders=('Order ID', 'nunique'),
    Average_Profit=('Profit', 'mean')
).sort_values(by='Total_Orders', ascending=False)

shipping_analysis
```

```python
monthly_sales = df.groupby(
    df['Order Date'].dt.to_period('M')
)['Sales'].sum()

monthly_sales = monthly_sales.reset_index()

monthly_sales.columns = ['Month', 'Total Sales']

monthly_sales
```

```python
subcategory_analysis = df.groupby('Sub-Category').agg(
    Total_Sales=('Sales', 'sum'),
    Total_Profit=('Profit', 'sum')
).sort_values(by='Total_Sales', ascending=False)

subcategory_analysis
```

```python
print("========== KEY BUSINESS INSIGHTS ==========")

# 1. Highest Sales Category
best_category = category_analysis['Total_Sales'].idxmax()
print("\n1. Highest Sales Category:", best_category)

# 2. Highest Sales Region
best_region = region_analysis['Total_Sales'].idxmax()
print("2. Highest Sales Region:", best_region)

# 3. Highest Sales Customer Segment
best_segment = segment_analysis['Total_Sales'].idxmax()
print("3. Highest Sales Segment:", best_segment)

# 4. Most Profitable Category
best_profit_category = category_analysis['Total_Profit'].idxmax()
print("4. Most Profitable Category:", best_profit_category)

# 5. Highest Sales Sub-category
best_subcategory = subcategory_analysis['Total_Sales'].idxmax()
print("5. Highest Sales Sub-category:", best_subcategory)

# 6. Most Used Shipping Mode
best_ship_mode = shipping_analysis['Total_Orders'].idxmax()
print("6. Most Used Shipping Mode:", best_ship_mode)

# 7. Highest Monthly Sales
best_month = monthly_sales.loc[
    monthly_sales['Total Sales'].idxmax()
]

print("7. Highest Sales Month:", best_month['Month'])
print("   Monthly Sales:", round(best_month['Total Sales'], 2))

# 8. Lowest Profit Category
lowest_profit_category = category_analysis['Total_Profit'].idxmin()
print("8. Lowest Profit Category:", lowest_profit_category)
```

```python
with pd.ExcelWriter('EDA_Analysis_Results.xlsx') as writer:

    missing_analysis.to_excel(
        writer, sheet_name='Missing Values'
    )

    category_analysis.to_excel(
        writer, sheet_name='Category Analysis'
    )

    region_analysis.to_excel(
        writer, sheet_name='Region Analysis'
    )

    segment_analysis.to_excel(
        writer, sheet_name='Segment Analysis'
    )

    outlier_df.to_excel(
        writer, sheet_name='Outlier Analysis'
    )

    monthly_sales.to_excel(
        writer, sheet_name='Monthly Sales', index=False
    )

print("EDA Results Exported Successfully!")
```
