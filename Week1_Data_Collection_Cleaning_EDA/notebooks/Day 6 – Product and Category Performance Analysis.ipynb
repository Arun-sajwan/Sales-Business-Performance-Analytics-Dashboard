### Import Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
df = pd.read_excel("sample_-_superstore.xls")
print("Dataset Loaded Successfully!")
```

### Check Dataset Columns

```python
print(df.shape)
print(df.columns.tolist())
df.head()
```

### Sales by Category

```python
category_sales = df.groupby('Category')['Sales'].sum().sort_values(ascending=False)

print(category_sales)
```

### Category Sales Visualization

```python
plt.figure(figsize=(8,5))

category_sales.plot(kind='bar')

plt.title('Sales by Product Category')
plt.xlabel('Category')
plt.ylabel('Total Sales')
plt.xticks(rotation=0)
plt.show()
```

### Sub-Category Performance Analysis

```python
subcategory_sales = df.groupby('Sub-Category')['Sales'].sum().sort_values(ascending=False)

print(subcategory_sales)
```

```python
plt.figure(figsize=(12,6))

subcategory_sales.plot(kind='bar', color='skyblue')

plt.title('Sales by Sub-Category')
plt.xlabel('Sub-Category')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### Top 10 Products by Sales

```python
product_sales = df.groupby('Product Name')['Sales'].sum()

top_products = product_sales.sort_values(ascending=False).head(10)

print(top_products)
```

```python
plt.figure(figsize=(12,6))

top_products.sort_values().plot(kind='barh')

plt.title('Top 10 Products by Sales')
plt.xlabel('Total Sales')
plt.ylabel('Product Name')
plt.tight_layout()
plt.show()
```

### Profit by Category

```python
category_profit = df.groupby('Category')['Profit'].sum().sort_values(ascending=False)

print(category_profit)
```

```python
plt.figure(figsize=(8,5))

category_profit.plot(kind='bar', color='orange')

plt.title('Profit by Product Category')
plt.xlabel('Category')
plt.ylabel('Total Profit')
plt.xticks(rotation=0)
plt.show()
```

### Lowest Sales Products

```python
bottom_products = product_sales.sort_values().head(10)

print(bottom_products)
```

```python
plt.figure(figsize=(12,6))

bottom_products.plot(kind='barh', color='salmon')

plt.title('Bottom 10 Products by Sales')
plt.xlabel('Total Sales')
plt.ylabel('Product Name')
plt.tight_layout()
plt.show()
```
