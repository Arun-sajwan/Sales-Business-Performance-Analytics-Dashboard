```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (10, 5)
```

```python
df = pd.read_excel("sample_-_superstore.xls")
print("Dataset Loaded Successfully!")
```

### Product Analysis

```python
product_sales = df.groupby('Product Name')['Sales'].sum()
print(product_sales.sort_values(ascending=False).head(10))
```

```python
product_profit = df.groupby('Product Name')['Profit'].sum()
print(product_profit.sort_values(ascending=False).head(10))
```

```python
product_quantity = df.groupby('Product Name')['Quantity'].sum()
print(product_quantity.sort_values(ascending=False).head(10))
```

```python
loss_products = product_profit[product_profit < 0]
print(loss_products.sort_values().head(10))
```

### Category Analysis

```python
category_analysis = df.groupby('Category').agg({
    'Sales': 'sum',
    'Profit': 'sum',
    'Quantity': 'sum'
})

print(category_analysis)

print(category_analysis.sort_values(
    by='Sales', ascending=False
))
```

### Customer Analysis

```python
customer_analysis = df.groupby('Customer Name').agg({
    'Sales': 'sum',
    'Profit': 'sum',
    'Quantity': 'sum',
    'Order ID': 'nunique'
})

print(customer_analysis.head())
```

```python
print(customer_analysis.sort_values(
    by='Sales', ascending=False
).head(10))
```

```python
print(customer_analysis.sort_values(
    by='Profit', ascending=False
).head(10))
```

### Regional Analysis

```python
region_analysis = df.groupby('Region').agg({
    'Sales': 'sum',
    'Profit': 'sum',
    'Quantity': 'sum'
})

print(region_analysis)
```

```python
print(region_analysis.sort_values(
    by='Sales', ascending=False
))
```

```python
print(region_analysis.sort_values(
    by='Profit', ascending=True
))
```
