### Import Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
```

### Load Dataset

```python
df = pd.read_excel("sample_-_superstore.xls")
print("Dataset Loaded Successfully!")
```

### Regional Revenue Analysis

```python
regional_revenue = df.groupby('Region')['Sales'].sum().reset_index()

regional_revenue = regional_revenue.sort_values(
    by='Sales', ascending=False
)

print(regional_revenue)
```

```python
plt.figure(figsize=(8,5))

sns.barplot(data=regional_revenue, x='Region', y='Sales')

plt.title('Regional Revenue Analysis')
plt.xlabel('Region')
plt.ylabel('Total Revenue')
plt.tight_layout()

plt.savefig('regional_revenue.png', dpi=300, bbox_inches='tight')
plt.show()
```

### Regional Profitability Analysis

```python
regional_profit = df.groupby('Region')['Profit'].sum().reset_index()

regional_profit = regional_profit.sort_values(
    by='Profit', ascending=False
)

print(regional_profit)
```

### Profit Visualization

```python
plt.figure(figsize=(8,5))

sns.barplot(data=regional_profit, x='Region', y='Profit')

plt.title('Regional Profitability Analysis')
plt.xlabel('Region')
plt.ylabel('Total Profit')
plt.tight_layout()

plt.savefig('regional_profit.png', dpi=300, bbox_inches='tight')
plt.show()
```

### Compare Revenue and Profit

```python
regional_performance = df.groupby('Region').agg({
    'Sales': 'sum',
    'Profit': 'sum'
}).reset_index()

print(regional_performance)
```

```python
regional_performance.plot(
    x='Region',
    y=['Sales', 'Profit'],
    kind='bar',
    figsize=(10,6)
)

plt.title('Regional Revenue vs Profit')
plt.xlabel('Region')
plt.ylabel('Amount')
plt.xticks(rotation=0)
plt.tight_layout()

plt.savefig('regional_revenue_vs_profit.png', dpi=300, bbox_inches='tight')
plt.show()
```

### Calculate Profit Margin

```python
regional_performance['Profit Margin (%)'] = (
    regional_performance['Profit'] /
    regional_performance['Sales']
) * 100

print(regional_performance)
```

### Identify Best and Lowest Performing Region

```python
best_revenue_region = regional_revenue.iloc[0]

best_profit_region = regional_profit.iloc[0]

lowest_profit_region = regional_profit.iloc[-1]

print("Highest Revenue Region:")
print(best_revenue_region)

print("\nHighest Profit Region:")
print(best_profit_region)

print("\nLowest Profit Region:")
print(lowest_profit_region)
```
