## Objective
Import the original Superstore Excel dataset using Pandas and inspect its dimensions, columns, data types, missing values, and sample records.

## 1. Import Libraries and Load Dataset

```python
import pandas as pd

file_path = "../../data/raw/sample_-_superstore.xls"
df = pd.read_excel(file_path, engine="xlrd")

print("Dataset loaded successfully.")
```

## 2. Dataset Dimensions

```python
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

## 3. Column Names

```python
print("Total columns:", len(df.columns))
for column in df.columns:
    print(column)
```

## 4. Data Types

```python
print(df.dtypes.to_string())
```

## 5. Missing Value Inspection

```python
missing_report = pd.DataFrame({
    "Missing_Count": df.isnull().sum(),
    "Missing_Percentage": (df.isnull().mean() * 100).round(2)
})

print(missing_report.to_string())
print("\nTotal missing values:", int(df.isnull().sum().sum()))
```

## 7. Initial Observations

- The dataset contains 10,194 rows and 21 columns.
- No missing values were detected during this inspection.
